# CEDS inventory and QA

Audits a CEDS release: file discovery, metadata integrity, time coverage,
source-id consistency, and truncated-download detection. Writes a provenance
manifest. **No aggregation happens here** — that is notebook 02.

Set `RELEASE` below to audit either release. The checks are identical for both,
which is what makes the old-vs-new comparison valid.

```python
from pathlib import Path
import sys

import pandas as pd

PROJECT_ROOT = Path.cwd()
if str(PROJECT_ROOT / "src") not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT / "src"))

from slcf_forcings.ceds import (
    MAIN_SPECIES,
    assert_all_readable,
    audit_collection,
    check_source_consistency,
    check_time_coverage,
    flag_truncated,
    select_files,
    write_manifest,
)

# "archive"    = CEDS-CMIP-2025-03-18 (reference; known CO/CO2/NOx gridding issue)
# "production" = CEDS-CMIP-2025-04-18 (current CMIP7 DECK source id)
RELEASE = "archive"

DATA_DIRS = {
    "archive": PROJECT_ROOT / "data" / "SLCF_main" / "old",
    "production": PROJECT_ROOT / "data" / "SLCF_main",
}
DATA_DIR = DATA_DIRS[RELEASE]
OUTPUT_DIR = PROJECT_ROOT / "output" / RELEASE
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Release:    {RELEASE}")
print(f"Data dir:   {DATA_DIR}")
print(f"Output dir: {OUTPUT_DIR}")
```

## 1. Metadata audit

Opens every file with `decode_times=False` — header reads only, no data.

```python
audit = audit_collection(DATA_DIR)
audit.to_csv(OUTPUT_DIR / "audit.csv", index=False)

print(f"{len(audit)} emissions files discovered\n")
print(
    audit.groupby(["stream", "species"])
    .agg(n_files=("file", "count"), total_gb=("size_bytes", lambda s: s.sum() / 1e9))
    .round(2)
    .to_string()
)
```

## 2. Integrity checks

Each check is a question with a verifiable answer, not a table to eyeball.

```python
unreadable = audit[~audit["readable"].fillna(False).astype(bool)]
print(f"Unreadable: {len(unreadable)} / {len(audit)}")
if not unreadable.empty:
    print(unreadable[["file", "size_bytes", "error"]].to_string(index=False))

suspect = flag_truncated(audit)
print(f"\nSuspiciously small for their chunk: {len(suspect)}")
if not suspect.empty:
    print(suspect[["file", "size_bytes"]].to_string(index=False))
```

```python
mismatched = check_source_consistency(audit)
print(f"source_id mismatches: {len(mismatched)}")
if not mismatched.empty:
    print(mismatched[["file", "source_id_filename", "source_id"]].to_string(index=False))

short = check_time_coverage(audit)
print(f"\nTime-coverage mismatches: {len(short)}")
if not short.empty:
    print(short[["file", "time_size", "expected_months"]].to_string(index=False))

ok = audit[audit["readable"].fillna(False).astype(bool)]
print(f"\nUnits: {sorted(ok['units'].dropna().unique())}")
print(f"Sector counts (surface): "
      f"{sorted(ok.loc[ok['stream'] == 'surface', 'n_sectors'].unique())}")
print(f"Dimension signatures: {sorted(ok['dimensions'].unique())}")
```

### Gate

Surface files are what notebook 02 aggregates, so they must all be readable.
The AIR stream is only needed for the pre-1970 extension (Phase 5), so it is
reported but does not block.

```python
surface = audit[audit["stream"] == "surface"]
assert_all_readable(surface, context=f"surface stream, {RELEASE}")
print(f"PASS: all {len(surface)} surface files readable")

air_bad = audit[(audit["stream"] == "AIR") & (~audit["readable"].fillna(False))]
if not air_bad.empty:
    print(
        f"\nWARNING: {len(air_bad)} AIR file(s) unreadable. Not required for the "
        "1970-2023 baseline; re-fetch before Phase 5."
    )
    print(air_bad[["file", "size_bytes"]].to_string(index=False))
```

## 3. Species and time coverage

Selection is by parsed time coverage, not chunk-string matching. CEDS chunk
boundaries move between releases — March splits CH4/N2O at 197001-201912,
April at 197001-199912 — so a literal match returns different species sets for
the two releases.

```python
WINDOWS = {
    "baseline_1970_2023": (1970, 2023),
    "working_2000_2023": (2000, 2023),
    "full_1750_2023": (1750, 2023),
}

for label, window in WINDOWS.items():
    try:
        files = select_files(DATA_DIR, MAIN_SPECIES, window)
        found = sorted({f.name.split("-em-")[0] for f in files})
        print(f"{label:22s} {len(files):3d} files, {len(found):2d} species")
    except ValueError as error:
        print(f"{label:22s} INCOMPLETE -- {error}")
```

```python
layout = (
    ok[ok["stream"] == "surface"]
    .sort_values(["species", "chunk_first_year"])
    [["species", "chunk_first_year", "chunk_last_year", "time_size", "n_sectors"]]
)
print(layout.to_string(index=False))
```

## 4. Provenance manifest

Closes the Phase 0 item: source id, version, coverage, licence and checksum per
input. Checksums are reused from the ESGF download manifest, which the download
wrapper already verified — that attests the bytes matched ESGF at download
time, a stronger claim than re-hashing local disk. Pass
`compute_checksums=True` to re-hash from disk instead.

```python
download_manifest = DATA_DIR / "download_manifest_gn.tsv"
if not download_manifest.exists():
    download_manifest = DATA_DIR.parent / "download_manifest_gn.tsv"

manifest = write_manifest(
    DATA_DIR,
    OUTPUT_DIR / "data_manifest.csv",
    download_manifest=download_manifest if download_manifest.exists() else None,
    compute_checksums=False,
)

print(f"Wrote {len(manifest)} rows to {OUTPUT_DIR / 'data_manifest.csv'}\n")
print(manifest[["file", "source_id", "first_year", "last_year", "n_months",
                "sha256_source"]].head(10).to_string(index=False))

missing = manifest["sha256"].isna().sum()
if missing:
    print(f"\n{missing} file(s) without a checksum — no matching ESGF manifest entry.")


bad = manifest[~manifest["readable"]]
if not bad.empty:
    print(f"\n{len(bad)} file(s) recorded as unreadable:")
    print(bad[["file", "size_bytes", "error"]].to_string(index=False))
```
