# Production-data transition and FaIR status

This notebook keeps the project status explicit: the archived old CEDS dataset remains the validated baseline, while the new production release and the FaIR analysis are separate follow-on steps.


```python
from pathlib import Path

repo = Path.cwd()
data_root = repo / "data" / "SLCF_main"
production_candidates = sorted(
    path for path in data_root.rglob("*.nc")
    if "old" not in path.parts and "areacella" not in path.name.lower()
)

print(f"Files under data/SLCF_main: {len(list(data_root.rglob('*')))}")
print(f"NetCDF files outside the old archive: {len(production_candidates)}")
for path in production_candidates[:10]:
    print(" -", path.relative_to(repo))
```

```python
import pandas as pd

baseline_path = repo / "output" / "preliminary" / "ceds_old_2000_2023_key_species_global.csv"
if baseline_path.exists():
    baseline = pd.read_csv(baseline_path)
    print("Old-data baseline is available.")
    print(baseline.head().to_string(index=False))
else:
    print("Old-data baseline file is not present yet.")
```

```python
if production_candidates:
    print("Production dataset is present; next step is a new-data aggregation step.")
else:
    print("No production CEDS files are present yet. The old archive remains the active reference until the new dataset is downloaded.")
```
